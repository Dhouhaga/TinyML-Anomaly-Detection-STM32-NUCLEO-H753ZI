```python
import numpy as np
import tensorflow as tf

# Raw sensor readings
raw_text = """                     53.83
                          53.49
                               53.31
                                    53.67
                                         53.67
                                              53.19
                                                   54.16
                                                        53.45
                                                             53.73
                                                                  53.47
                                                                       53.45
                                                                            53.92
 53.23
      53.37
           53.57
                53.71
                     53.53
                          53.55
                               53.51
                                    53.21
                                         53.79
                                              53.35
                                                   53.43
                                                        53.37
                                                             53.75
                                                                  53.47
                                                                       53.77
                                                                            53.59
 53.57
      53.59
           53.67
                53.49
                     53.75
                          53.73
                               53.63
                                    53.53
                                         53.37
                                              53.71
                                                   53.94
                                                        53.75
                                                             53.11
                                                                  53.45
                                                                       53.15
                                                                            53.71
 53.43
      53.55
           53.87
                53.96
                     53.79
                          54.12
                               53.49
                                    53.33
                                         53.65
                                              53.55
                                                   53.57
                                                        53.92
                                                             52.96
                                                                  53.92
                                                                       53.73
                                                                            53.71
 53.75
      53.61
           53.89
                53.51
                     53.53
                          53.25
                               53.65
                                    53.83
                                         53.51
                                              53.21
                                                   54.08
                                                        54.00
                                                             53.37
                                                                  53.27
                                                                       53.61
                                                                            52.94
 53.75
      53.47
           53.39
                53.35
                     53.69
                          53.71
                               53.27
                                    53.59
                                         53.89
                                              53.71
                                                   53.92
                                                        54.06
                                                             53.15
                                                                  54.04
                                                                       53.47
                                                                            53.27
 53.89
      53.85
           53.49
                53.51
                     53.67
                          53.27
                               53.96
                                    53.69
                                         53.57
                                              53.96
                                                   53.94
                                                        53.51
                                                             53.59
                                                                  53.79
                                                                       53.96
                                                                            53.63
 53.65
      53.75
           54.06
                53.23
                     53.65
                          53.83
                               53.15
                                    53.77
                                         53.81
                                              53.53
                                                   53.45
                                                        53.53
                                                             53.79
                                                                  53.69
                                                                       54.08
                                                                            53.75
 53.51
      53.33
           53.57
                53.25
                     53.73
                          53.65
                               54.06
                                    53.31
                                         53.41
                                              53.25
                                                   54.02
                                                        53.51
                                                             53.57
                                                                  53.73
                                                                       54.72
                                                                            53.81
 54.20
      53.87
           53.15
                53.45
                     53.81
                          53.69
                               54.06
                                    53.69
                                         54.50
                                              53.41
                                                   53.41
                                                        53.55
                                                             53.79
                                                                  53.98
                                                                       53.81
                                                                            53.67
 53.19
      53.35
           53.63
                53.61
                     53.89
                          53.59
                               53.41
                                    53.92
                                         53.31
                                              53.43
                                                   53.59
                                                        53.85
                                                             53.51
                                                                  53.27
                                                                       53.19
                                                                            53.57
 53.69
      53.55
           53.37
                53.79
                     53.41
                          54.22
                               53.98
                                    53.51
                                         53.37
                                              53.45
                                                   54.24
                                                        53.65
                                                             53.27
                                                                  53.47
                                                                       53.61
                                                                            53.31
 54.08
      53.55
           54.20
                53.21
                     54.30
                          53.53
                               53.31
                                    54.04
                                         53.63
                                              53.45
                                                   53.77
                                                        53.98
                                                             53.92
                                                                  53.63
                                                                       53.57
                                                                            53.47
 53.92
      53.69
           54.14
                53.27
                     53.75
                          53.39
                               53.41
                                    53.31
                                         53.47
                                              53.65
                                                   53.92
                                                        53.45
                                                             53.79
                                                                  53.85
                                                                       53.39
                                                                            53.63
 53.98
      54.10
           54.12
                53.63
                     53.79
                          53.49
                               53.21
                                    53.96
                                         53.92
                                              53.49
                                                   53.59
                                                        53.79
                                                             53.85
                                                                  53.13
                                                                       53.51
                                                                            53.83
 53.61
      53.59
           54.26
                53.79
                     53.81
                          53.37
                               53.61
                                    53.39
                                         54.02
                                              53.41
                                                   53.92
                                                        54.10
                                                             53.67
                                                                  53.51
                                                                       53.81
                                                                            53.45
 53.27
      53.41
           54.06
                53.92
                     53.13
                          53.69
                               53.71
                                    53.83
                                         53.85
                                              53.69
                                                   54.12
                                                        53.63
                                                             53.75
                                                                  53.45
                                                                       53.57
                                                                            53.35
 53.63
      53.59
           53.77
                53.87
                     53.33
                          54.00
                               53.59
                                    53.73
                                         53.19
                                              53.73
                                                   53.59
                                                        53.57
                                                             53.96
                                                                  53.43
                                                                       53.69
                                                                            53.47
 54.02
      53.87
           53.79
                53.25
                     53.75
                          53.19
                               53.92
                                    53.35
                                         53.11
                                              53.53
                                                   52.98
                                                        53.79
                                                             54.06
                                                                  53.65
                                                                       53.31
                                                                            53.77
 54.12
      53.59
           53.35
                53.45
                     53.53
                          53.53
                               53.45
                                    53.96
                                         53.75
                                              54.22
                                                   53.49
                                                        54.10
                                                             53.33
                                                                  53.41
                                                                       53.15
                                                                            53.83
 53.65
      53.92
           53.87
                53.37
                     53.41
                          53.77
                               53.79
                                    53.23
                                         53.69
                                              53.69
                                                   53.85
                                                        53.65
                                                             53.57
                                                                  53.47
                                                                       53.96
                                                                            53.73
 53.89
      53.87
           53.21
                53.11
                     53.47
                          53.53
                               53.57
                                    53.25
                                         53.41
                                              53.31
                                                   54.20
                                                        53.92
                                                             53.41
                                                                  54.62
                                                                       53.94
                                                                            53.53
 53.83
      53.79
           53.21
                53.77
                     53.21
                          53.51
                               53.65
                                    53.83
                                         54.12
                                              53.39
                                                   54.22
                                                        53.71
                                                             53.79
                                                                  53.81
                                                                       53.53
                                                                            53.47
 53.89
      53.57
           53.63
                53.25
                     53.27
                          53.49
                               53.65
                                    53.75
                                         53.61
                                              53.57
                                                   53.53
                                                        53.33
                                                             54.02
                                                                  53.71
                                                                       53.79
                                                                            53.43
 53.43
      53.98
           53.47
                53.89
                     53.45
                          53.67
                               53.51
                                    53.83
                                         53.21
                                              53.61
                                                   53.55
                                                        54.08
                                                             53.75
                                                                  53.59
                                                                       54.06
                                                                            53.55
 54.14
      53.07
           53.59
                53.81
                     53.45
                          53.51
                               53.41
                                    54.08
                                         53.87
                                              53.29
                                                   53.79
                                                        53.65
                                                             53.55
                                                                  53.79
                                                                       53.67
                                                                            54.32
 53.92
      54.32
           53.73
                53.87
                     53.61
                          53.73
                               53.41
                                    53.57
                                         53.75
                                              53.59
                                                   53.75
                                                        53.85
                                                             53.79
                                                                  53.96
                                                                       53.61
                                                                            53.69
 53.55
      53.94
           53.49
                53.51
                     53.51
                          53.85
                               53.37
                                    53.27
                                         53.31
                                              53.53
                                                   53.57
                                                        54.04
                                                             53.29
                                                                  53.19
                                                                       53.41
                                                                            53.49
 53.25
      53.79
           53.57
                54.36
                     53.59
                          53.57
                               53.35
                                    53.47
                                         53.69
                                              53.67
                                                   53.55
                                                        53.92
                                                             53.37
                                                                  53.51
                                                                       53.45
                                                                            53.39
 53.55
      53.53
           53.96
                54.06
                     53.75
                          53.75
                               53.65
                                    53.45
                                         53.29
                                              54.02
                                                   53.53
                                                        53.35
                                                             53.31
                                                                  54.00
                                                                       53.11
                                                                            54.16
 53.92
      53.55
           53.98
                53.71
                     53.63
                          53.81
                               53.63
                                    53.67
                                         53.53
                                              53.55
                                                   54.76
                                                        53.23
                                                             53.75
                                                                  53.21
                                                                       53.39
                                                                            53.25
 53.94
      53.77
           54.08
                54.04
                     53.81
                          53.39
                               53.77
                                    53.69
                                         53.69
                                              53.25
                                                   53.53
                                                        53.92
                                                             53.92
                                                                  53.81
                                                                       53.83
                                                                            53.75
 53.33
      53.43
           54.00
                53.65
                     53.57
                          53.43
                               53.65
                                    54.06
                                         53.85
                                              53.98
                                                   54.00
                                                        53.75
                                                             54.38
                                                                  53.73
                                                                       53.43
                                                                            53.57
 53.47
      53.77
           53.85
                53.37
                     53.75
                          53.63
                               53.67
                                    53.75
                                         53.45
                                              53.27
                                                   53.71
                                                        53.59
                                                             53.75
                                                                  53.92
                                                                       53.43
                                                                            53.51
 53.51
      53.35
           54.72
                53.83
                     53.83
                          53.43
                               53.59
                                    53.45
                                         54.20
                                              53.41
                                                   53.87
                                                        53.71
                                                             53.49
                                                                  54.02
                                                                       54.10
                                                                            53.73
 53.57
      53.19
           53.53
                53.45
                     53.57
                          53.39
                               53.55
                                    53.83
                                         53.65
                                              54.14
                                                   54.14
                                                        53.92
                                                             53.29
                                                                  53.94
                                                                       53.79
                                                                            53.49
 53.65
      54.02
           53.75
                53.51
                     53.75
                          53.59
                               53.65
                                    53.71
                                         53.51
                                              53.81
                                                   53.37
                                                        53.47
                                                             53.57
                                                                  53.53
                                                                       53.61
                                                                            53.49
 53.11
      53.37
           53.25
                53.61
                     53.65
                          53.35
                               54.00
                                    53.27
                                         53.47
                                              53.67
                                                   53.49
                                                        53.89
                                                             53.55
                                                                  53.92
                                                                       53.51
                                                                            53.63
 53.65
      53.51
           53.65
                53.53
                     53.49
                          53.51
                               53.53
                                    53.45
                                         53.11
                                              53.39
                                                   53.87
                                                        53.35
                                                             53.17
                                                                  53.92
                                                                       53.57
                                                                            53.51
 53.41
      53.85
           53.41
                53.43
                     53.51
                          53.51
                               53.33
                                    53.65
                                         54.16
                                              53.75
                                                   53.47
                                                        53.59
                                                             53.57
                                                                  53.67
                                                                       54.04
                                                                            53.49
 53.73
      53.92
           53.51
                53.43
                     54.06
                          53.57
                               54.16
                                    53.67
                                         53.65
                                              53.69
                                                   53.75
                                                        53.65
                                                             53.45
                                                                  53.15
                                                                       53.71
                                                                            53.83
 53.41
      53.43
           53.92
                53.43
                     53.49
                          53.41
                               53.41
                                    52.72
                                         53.51
                                              53.85
                                                   53.87
                                                        53.75
                                                             53.11
                                                                  53.29
                                                                       54.04
                                                                            53.21
 53.47
      53.39
           53.27
                53.45
                     53.43
                          53.67
                               53.43
                                    53.55
                                         53.27
                                              53.85
                                                   53.71
                                                        53.94
                                                             53.23
                                                                  53.57
                                                                       53.47
                                                                            53.65
 54.10
      53.35
           53.75
                53.55
                     53.67
                          53.43
                               53.59
                                    53.47
                                         53.75
                                              53.75
                                                   53.55
                                                        53.51
                                                             53.53
                                                                  53.35
                                                                       53.41
                                                                            53.49
 53.43
      53.98
           53.75
                53.51
                     53.57
                          53.92
                               54.06
                                    53.23
                                         53.57
                                              53.81
                                                   53.92
                                                        53.23
                                                             54.08
                                                                  53.37
                                                                       53.89
                                                                            53.73
 54.10
      53.31
           53.67
                53.59
                     52.54
                          53.31
                               53.71
                                    53.63
                                         53.75
                                              53.94
                                                   53.35
                                                        54.22
                                                             53.31
                                                                  53.65
                                                                       53.96
                                                                            53.43
 53.43
      53.37
           53.17
                53.75
                     53.94
                          53.21
                               53.63
                                    53.11
                                         53.63
                                              53.98
                                                   53.67
                                                        53.71
                                                             53.79
                                                                  53.37
                                                                       53.49
                                                                            53.41
 53.43
      53.75
           53.39
                53.43
                     53.57
                          53.71
                               53.35
                                    53.27
                                         53.89
                                              54.06
                                                   53.59
                                                        53.81
                                                             53.33
                                                                  54.06
                                                                       53.43
                                                                            53.85
 53.49
      53.49
           54.06
                53.35
                     53.92
                          53.41
                               53.45
                                    53.21
                                         53.87
                                              53.49
                                                   53.65
                                                        53.96
                                                             53.55
                                                                  53.11
                                                                       53.65
                                                                            53.75
 54.08
      53.11
           53.94
                53.75
                     53.92
                          53.53
                               53.89
                                    53.83
                                         54.06
                                              53.43
                                                   53.55
                                                        53.77
                                                             53.92
                                                                  53.87
                                                                       53.65
                                                                            53.92
 53.43
      54.16
           53.47
                53.11
                     53.73
                          53.33
                               53.69
                                    53.39
                                         53.43
                                              53.59
                                                   53.63
                                                        53.81
                                                             53.67
                                                                  53.43
                                                                       53.59
                                                                            53.55
 53.51
      54.16
           53.69
                53.57
                     53.98
                          53.75
                               53.19
                                    53.77
                                         53.63
                                              53.49
                                                   52.58
                                                        53.83
                                                             54.02
                                                                  53.39
                                                                       53.43
                                                                            53.57
 53.59
      53.75
           53.55
                54.24
                     53.65
                          54.06
                               53.53
                                    53.73
                                         53.33
                                              53.92
                                                   54.12
                                                        53.53
                                                             53.79
                                                                  53.43
                                                                       53.63
                                                                            53.43
 53.59
      53.67
           53.31
                53.43
                     53.77
                          53.79
                               53.43
                                    53.47
                                         53.43
                                              53.25
                                                   53.35
                                                        53.45
                                                             53.63
                                                                  53.31
                                                                       53.35
                                                                            54.06
 53.63
      53.53
           53.57
                53.37
                     53.67
                          53.75
                               53.63
                                    53.55
                                         53.92
                                              53.53
                                                   53.45
                                                        53.41
                                                             53.33
                                                                  53.63
                                                                       53.75
                                                                            53.53
 53.83
      53.47
           52.78
                53.41
                     53.63
                          53.51
                               53.23
                                    53.49
                                         53.47
                                              53.65
                                                   53.57
                                                        53.92
                                                             53.43
                                                                  53.41
                                                                       53.37
                                                                            53.96
 53.55
      53.29
           53.59
                53.61
                     53.31
                          53.25
                               53.43
                                    53.75
                                         53.59
                                              53.59
                                                   53.81
                                                        53.85
                                                             54.12
                                                                  53.31
                                                                       53.92
                                                                            53.89
 53.73
      53.75
           53.69
                53.63
                     54.14
                          53.83
                               53.73
                                    53.45
                                         53.37
                                              53.41
                                                   53.65
                                                        53.55
                                                             53.89
                                                                  53.83
                                                                       53.69
                                                                            53.37
 54.00
      53.43
           53.59
                54.04
                     53.49
                          53.43
                               53.43
                                    53.75
                                         53.43
                                              53.57
                                                   53.43
                                                        53.79
                                                             53.85
                                                                  53.31
                                                                       54.04
                                                                            54.00
 53.75
      53.19
           53.89
                53.37
                     53.15
                          53.71
                               53.57
                                    53.33
                                         53.73
                                              53.75
                                                   53.65
                                                        53.55
                                                             53.55
                                                                  53.71
                                                                       53.41
                                                                            53.35
 53.87
      53.63
           53.19
                53.43
                     53.98
                          53.77
                               53.49
                                    54.02
                                         53.29
                                              53.73
                                                   53.31
                                                        53.94
                                                             53.71
                                                                  53.41
                                                                       53.61
                                                                            53.17
 53.41
      54.00
           53.21
                53.75
                     53.49
                          53.41
                               53.51
                                    53.71
                                         53.96
                                              53.83
                                                   53.49
                                                        53.53
                                                             53.63
                                                                  53.89
                                                                       53.17
                                                                            53.98
 53.33
      53.73
           53.83
                53.51
                     54.38
                          53.47
                               53.23
                                    53.49
                                         53.85
                                              53.53
                                                   53.51
                                                        53.83
                                                             53.47
                                                                  53.19
                                                                       53.96
                                                                            54.06
 53.73
      53.29
           54.06
                53.11
                     53.87
                          53.73
                               53.75
                                    53.73
                                         53.19
                                              53.89
                                                   53.11
                                                        53.85
                                                             53.92
                                                                  53.51
                                                                       53.65
                                                                            53.75
 53.96
      53.37
           53.31
                53.96
                     53.19
                          53.69
                               53.27
                                    53.61
                                         53.33
                                              53.45
                                                   53.77
                                                        53.39
                                                             53.55
                                                                  53.39
                                                                       53.33
                                                                            53.79
 53.83
      53.89
           53.98
                53.59
                     54.06
                          53.51
                               53.41
                                    53.57
                                         53.19
                                              53.47
                                                   53.19
                                                        53.89
                                                             53.49
                                                                  53.31
                                                                       53.71
                                                                            53.37
 53.57
      54.85
           53.25
                53.63
                     53.69
                          53.96
                               53.69
                                    53.71
                                         53.96
                                              53.92
                                                   53.53
                                                        53.81
                                                             53.85
                                                                  53.19
                                                                       53.27
                                                                            53.61
 53.51
      53.53
           53.15
                53.43
                     53.43
                          53.98
                               53.87
                                    53.73
                                         53.41
                                              53.49
                                                   53.51
                                                        53.21
                                                             53.27
                                                                  53.94
                                                                       53.57
                                                                            53.27
 53.11
      53.47
           53.83
                53.43
                     53.96
                          53.47
                               54.06
                                    53.47
                                         53.19
                                              53.57
                                                   53.87
                                                        53.49
                                                             53.43
                                                                  53.37
                                                                       53.69
                                                                            53.89
 53.31
      53.75
           53.83
                53.19
                     53.55
                          53.75
                               53.45
                                    53.83
                                         54.10
                                              53.29
                                                   53.31
                                                        53.57
                                                             53.75
                                                                  53.55
                                                                       53.59
                                                                            53.77
 54.08
      53.83
           53.92
                54.04
                     53.57
                          53.41
                               54.08
                                    53.27
                                         53.39
                                              53.73
                                                   53.63
                                                        53.59
                                                             53.49
                                                                  53.94
                                                                       53.47
                                                                            53.69
 53.21
      53.45
           53.47
                53.45
                     53.75
                          53.53
                               53.51
                                    53.11
                                         53.41
                                              53.19
                                                   53.94
                                                        53.47
                                                             54.18
                                                                  54.04
                                                                       53.29
                                                                            53.37
 53.17
      53.59
           54.04
                53.43
                     53.94
                          53.63
                               53.59
                                    53.61
                                         53.79
                                              53.75
                                                   53.55
                                                        53.29
                                                             53.75
                                                                  53.89
                                                                       54.00
                                                                            53.96
 53.29
      53.96
           53.92
                53.41
                     53.37
                          53.19
                               53.96
                                    53.75
                                         52.82
                                              54.08
                                                   53.49
                                                        53.43
                                                             53.61
                                                                  53.51
                                                                       53.75
                                                                            53.27
 53.41
      54.00
           53.49
                53.69
                     53.45
                          53.43
                               53.53
                                    53.69
                                         53.85
                                              53.25
                                                   54.02
                                                        53.19
                                                             53.13
                                                                  53.51
                                                                       53.61
                                                                            54.00
 53.57
      53.61
           54.02
                53.51
                     53.73
                          53.59
                               53.81
                                    53.47
                                         54.14
                                              53.37
                                                   54.64
                                                        53.43
                                                             53.83
                                                                  53.63
                                                                       53.31
                                                                            53.43
 53.77
      53.92
           53.96
                53.53
                     53.45
                          53.69
                               53.85
                                    53.83
                                         53.47
                                              53.85
                                                   53.83
                                                        52.54
                                                             53.31
                                                                  53.75
                                                                       53.55
                                                                            53.59
 53.17
      53.63
           53.27
                53.27
                     53.45
                          53.51
                               53.92
                                    53.37
                                         53.85
                                              53.92
                                                   53.81
                                                        53.51
                                                             53.57
                                                                  53.92
                                                                       53.43
                                                                            53.51
 54.00
      53.77
           53.27
                54.00
                     53.83
                          53.55
                               53.15
                                    53.69
                                         52.62
                                              53.63
                                                   53.33
                                                        53.75
                                                             53.49
                                                                  53.92
                                                                       53.31
                                                                            53.59
 53.43
      53.63
           53.63
                53.75
                     53.92
                          53.31
                               53.41
                                    53.41
                                         53.11
                                              53.47
                                                   53.96
                                                        53.43
                                                             53.51
                                                                  53.73
                                                                       53.27
                                                                            53.25
 53.57
      53.75
           53.83
                53.39
                     53.92
                          53.57
                               53.61
                                    53.65
                                         53.33
                                              53.67
                                                   53.27
                                                        53.63
                                                             53.75
                                                                  54.00
                                                                       53.67
                                                                            53.45
 53.45
      53.49
           53.75
                53.67
                     53.59
                          53.39
                               53.73
                                    53.31
                                         53.75
                                              53.75
                                                   53.89
                                                        53.75
                                                             53.47
                                                                  53.98
                                                                       53.59
                                                                            53.47
 53.75
      54.22
           53.53
                53.67
                     53.57
                          53.92
                               53.47
                                    53.39
                                         53.31
                                              53.49
                                                   53.51
                                                        54.06
                                                             53.41
                                                                  53.47
                                                                       53.33
                                                                            53.59
 53.33
      53.67
           53.37
                54.02
                     53.85
                          53.83
                               53.51
                                    53.47
                                         53.69
                                              52.94
                                                   53.73
                                                        53.47
                                                             53.65
                                                                  53.92
                                                                       53.77
                                                                            53.31
 53.98
      53.75
           53.55
                53.75
                     53.75
                          53.47
                               53.73
                                    53.35
                                         53.51
                                              53.69
                                                   53.94
                                                        53.73
                                                             53.59
                                                                  53.51
                                                                       53.83
                                                                            53.75
 53.59
      53.31
           54.26
                53.11
                     53.57
                          53.43
                               53.77
                                    53.47
                                         53.31
                                              53.31
                                                   53.45
                                                        53.75
                                                             53.75
                                                                  53.27
                                                                       53.92
                                                                            53.61
 53.43
      53.43
           53.51
                53.67
                     53.29
                          53.53
                               53.71
                                    53.94
                                         53.27
                                              53.53
                                                   54.06
                                                        54.06
                                                             53.45
                                                                  53.29
                                                                       53.11
                                                                            53.63
 53.92
      53.83
           53.57
                53.31
                     53.31
                          53.43
                               53.29
                                    53.29
                                         53.55
                                              53.33
                                                   53.79
                                                        53.79
                                                             53.43
                                                                  53.19
                                                                       53.69
                                                                            53.51
 54.08
      52.94
           53.25
                53.51
                     53.19
                          53.25
                               53.67
                                    53.33
                                         54.04
                                              54.08
                                                   53.65
                                                        52.60
                                                             53.15
                                                                  53.75
                                                                       53.57
                                                                            53.41
 53.43
      53.81
           53.87
                53.31
                     53.37
                          53.51
                               53.53
                                    52.70
                                         53.47
                                              53.73
                                                   53.35
                                                        53.75
                                                             53.85
                                                                  53.57
                                                                       53.71
                                                                            53.35
 53.41
      53.19
           53.69
                53.57
                     53.98
                          53.63
                               53.27
                                    53.75
                                         53.49
                                              53.55
                                                   54.32
                                                        53.31
                                                             53.63
                                                                  54.20
                                                                       53.29
                                                                            53.87
 53.57
      53.89
           53.92
                53.35
                     53.47
                          53.45
                               53.19
                                    53.53
                                         53.79
                                              53.55
                                                   53.49
                                                        53.55
                                                             53.83
                                                                  53.61
                                                                       53.75
                                                                            53.43
 53.59
      53.59
           54.12
                53.51
                     53.81
                          53.57
                               53.43
                                    53.41
                                         53.31
                                              53.75
                                                   53.47
                                                        53.65
                                                             53.33
                                                                  53.96
                                                                       53.67
                                                                            54.08
 53.11
      54.18
           53.41
                54.08
                     53.67
                          53.85
                               53.55
                                    53.75
                                         53.47
                                              53.43
                                                   53.35
                                                        53.35
                                                             54.10
                                                                  53.59
                                                                       53.65
                                                                            54.24
 54.06
      53.55
           54.06
                53.96
                     53.55
                          53.57
                               53.69
                                    53.63
                                         53.41
                                              53.83
                                                   53.55
                                                        54.14
                                                             53.33
                                                                  53.31
                                                                       53.39
                                                                            53.43
 53.69
      53.19
           53.57
                53.98
                     53.67
                          53.49
                               53.27
                                    53.59
                                         53.75
                                              53.47
                                                   53.49
                                                        53.37
                                                             53.75
                                                                  53.41
                                                                       53.79
                                                                            54.08
 53.19
      53.19
           53.35
                53.75
                     53.77
                          54.04
                               53.59
                                    52.90
                                         53.19
                                              53.33
                                                   53.63
                                                        53.35
                                                             53.21
                                                                  53.27
                                                                       53.33
                                                                            53.73
 53.83
      53.75
           53.25
                53.45
                     53.29
                          53.89
                               53.45
                                    54.02
                                         53.87
                                              53.65
                                                   54.08
                                                        53.57
                                                             53.73
                                                                  53.53
                                                                       53.31
                                                                            53.75
 53.43
      53.63
           53.39
                53.57
                     53.71
                          53.63
                               53.89
                                    53.89
                                         54.00
                                              53.47
                                                   53.75
                                                        53.49
                                                             53.27
                                                                  53.92
                                                                       53.59
                                                                            53.92
 53.15
      53.55
           53.75
                53.35
                     53.75
                          53.51
                               53.43
                                    53.77
                                         53.59
                                              53.53
                                                   53.85
                                                        53.77
                                                             53.53
                                                                  53.43
                                                                       53.49
                                                                            53.92
 53.33
      53.75
           53.27
                53.59
                     53.45
                          53.49
                               53.57
                                    53.67
                                         53.75
                                              53.92
                                                   53.59
                                                        53.43
                                                             53.33
                                                                  53.53
                                                                       53.47
                                                                            53.98
 53.11
      53.51
           53.69
                53.89
                     53.69
                          53.67
                               54.16
                                    53.92
                                         53.79
                                              53.37
                                                   53.73
                                                        53.67
                                                             53.31
                                                                  53.15
                                                                       53.83
                                                                            53.29
 53.43
      53.75
           53.53
                53.11
                     53.69
                          53.33
                               54.28
                                    53.51
                                         53.51
                                              53.92
                                                   53.43
                                                        53.61
                                                             53.43
                                                                  53.33
                                                                       53.89
                                                                            53.96
 53.61
      53.17
           53.77
                54.00
                     53.47
                          53.79
                               53.79
                                    53.75
                                         53.67
                                              53.55
                                                   53.96
                                                        53.77
                                                             53.61
                                                                  53.31
                                                                       53.98
                                                                            53.57"""

# Parse
temps = [float(x.strip()) for x in raw_text.strip().split('\n') if x.strip()]
temps = np.array(temps, dtype=np.float32)
print(f"Samples: {len(temps)}, Mean: {temps.mean():.2f}, Std: {temps.std():.2f}")

# Normalize
mean, std = temps.mean(), temps.std()
temps_norm = (temps - mean) / std

# Create sliding windows of 10 samples
window = 10
X = np.array([temps_norm[i:i+window] for i in range(len(temps_norm)-window)])
print(f"Windows: {X.shape}")

# Train autoencoder
inp = tf.keras.Input(shape=(window,))
x   = tf.keras.layers.Dense(8, activation='relu')(inp)
x   = tf.keras.layers.Dense(4, activation='relu')(x)   # bottleneck
x   = tf.keras.layers.Dense(8, activation='relu')(x)
out = tf.keras.layers.Dense(window)(x)
model = tf.keras.Model(inp, out)
model.compile(optimizer='adam', loss='mse')
model.fit(X, X, epochs=50, batch_size=32, validation_split=0.1, verbose=1)

# Threshold
preds = model.predict(X)
errors = np.mean((preds - X)**2, axis=1)
threshold = np.percentile(errors, 95)
print(f"\nThreshold: {threshold:.6f}")
print(f"Mean: {mean:.4f}, Std: {std:.4f}  <- save these for firmware")

# Convert to TFLite int8
def rep_data():
    for i in range(min(200, len(X))):
        yield [X[i:i+1]]

converter = tf.lite.TFLiteConverter.from_keras_model(model)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
converter.representative_dataset = rep_data
converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTINS_INT8]
converter.inference_input_type  = tf.int8
converter.inference_output_type = tf.int8
tflite = converter.convert()

with open('temp_anomaly_model.tflite', 'wb') as f:
    f.write(tflite)
print(f"Model size: {len(tflite)/1024:.1f} KB")
```
